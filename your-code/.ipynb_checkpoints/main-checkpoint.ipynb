# Challenge 2 - Hypothesis Tests

In this section of the lab, we will test whether the hourly wage of all hourly workers is significantly different from $30/hr. Import the correct one sample test function from scipy and perform the hypothesis test for a 95% two sided confidence interval.

```python
hourly = salaries[salaries['Salary or Hourly'] == 'Hourly']  # Filter only hourly workers

hourly_wages = hourly['Hourly Rate'] # Extract their hourly wages

t_stat, p_value = ttest_1samp(hourly_wages, 30)  # Perform a one-sample t-test against a population mean of 30

# Show the results:
print("T-statistic:", t_stat)
print("P-value:", p_value)

#The p-value is extremely small (much less than 0.05), so we reject the null hypothesis. This means the average hourly wage is significantly different from $30/hour.
```

# Challenge 4 -  Hypothesis Tests of Proportions

## This challenge is OPTIONAL

Another type of one sample test is a hypothesis test of proportions. In this test, we examine whether the proportion of a group in our sample is significantly different than a fraction. 

You can read more about one sample proportion tests [here](http://sphweb.bumc.bu.edu/otlt/MPH-Modules/BS/SAS/SAS6-CategoricalData/SAS6-CategoricalData2.html).

In the cell below, use the `proportions_ztest` function from `statsmodels` to perform a hypothesis test that will determine whether the number of hourly workers in the City of Chicago is significantly different from 25% at the 95% confidence level.

```python
from statsmodels.stats.proportion import proportions_ztest  #Import the function for proportion z-test

total_employees = len(salaries) # Count the total number of employees

hourly_count = len(salaries[salaries['Salary or Hourly'] == 'Hourly']) # Count the number of hourly workers

z_stat, p_value = proportions_ztest(count=hourly_count, nobs=total_employees, value=0.25) # Run the z-test to compare the proportion of hourly workers to 25%
# count = number of hourly workers
# nobs = total number of workers
# value = 0.25 means we are testing against 25%

# Print the results:
print("Total employees:", total_employees)
print("Hourly workers:", hourly_count)
print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Interpret the result:
if p_value < 0.05:
    print("The proportion of hourly workers is significantly different from 25%.")
else:
    print("There is no significant difference from 25%.")

#We tested whether the proportion of hourly workers (24.17%) is different from 25%. The z-statistic is -3.51 and the p-value is 0.0004, which is lower than 0.05. Therefore, we reject the null hypothesis and conclude that the proportion of hourly workers is significantly different from 25%.

```
