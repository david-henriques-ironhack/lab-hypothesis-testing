# Challenge 2 - Hypothesis Tests

In this section of the lab, we will test whether the hourly wage of all hourly workers is significantly different from $30/hr. Import the correct one sample test function from scipy and perform the hypothesis test for a 95% two sided confidence interval.

```python
wages = salaries \
    .loc[salaries['Salary or Hourly'] == 'Hourly', 'Hourly Rate']
```

```python
wages.describe()
```

```python
sns.histplot(wages);
```

H0: mean wage = 30
Ha: mean wage != 30

population std is unknown - need to use a one-sample t-test
two-sided confidence interval because testing for equality

Apha = 0.05

```python
ttest_1samp(wages, 30)
```

p-value = 0

p-value < alpha, thus, H0 is rejected

the population mean salary is significantly different from zero

```python
print(f'Mean: {round(wages.mean(), 2)}')
print(f'N: {wages.count()}')
print(f'Sample std: {round(wages.std(), 2)}')
print(f'SE: {round(wages.sem(), 3)}')
```

```python
t.interval(
    confidence = .95,
    df = wages.count() - 1,
    loc = wages.mean(),
    scale = wages.sem()
)
```

# Challenge 4 -  Hypothesis Tests of Proportions

## This challenge is OPTIONAL

Another type of one sample test is a hypothesis test of proportions. In this test, we examine whether the proportion of a group in our sample is significantly different than a fraction. 

You can read more about one sample proportion tests [here](http://sphweb.bumc.bu.edu/otlt/MPH-Modules/BS/SAS/SAS6-CategoricalData/SAS6-CategoricalData2.html).

In the cell below, use the `proportions_ztest` function from `statsmodels` to perform a hypothesis test that will determine whether the number of hourly workers in the City of Chicago is significantly different from 25% at the 95% confidence level.

```python
h_workers_count = salaries.loc[salaries['Salary or Hourly'] == 'Hourly', 'Salary or Hourly'].count()
t_workers_count = salaries.loc[:, 'Salary or Hourly'].count()
sample_p = h_workers_count / t_workers_count

print(f'Sample proportion of hourly workers: {round(sample_p, 3)}')
```

```python
# is the sample size large enough for normal approximation?

p_zero = .25

print(f'n * p_zero = {t_workers_count * p_zero}')
print(f'n * (1 - p_zero) = {t_workers_count * (1 - p_zero)}')

print('\n')

if ((t_workers_count * p_zero) > 10) & ((t_workers_count * (1 - p_zero)) > 10):
    print('Sample size is large enough')
else:
    print('Sample size is NOT large enough')
```

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
proportions_ztest(
    h_workers_count,
    t_workers_count,
    p_zero)
```

The sample proportion lies 3.5 SEs below P0
p-value < alpha, thus Ho is rejected

```python
from statsmodels.stats.proportion import proportion_confint
```

```python
proportion_confint(count=h_workers_count, 
                                  nobs=t_workers_count, 
                                  alpha=0.05, 
                                  method='normal')
```

As concluded above, 0.25 is outside the confidence interval rate uder CL of 95%
